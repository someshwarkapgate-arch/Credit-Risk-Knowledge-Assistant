```python
documents = load_all_pdfs(DOCUMENT_DIR)

print("Total page-level documents:", len(documents))
```

```python
print(type(documents))
print(type(documents[0]))

print("\nMetadata:")
print(documents[0].metadata)

print("\nContent:")
print(documents[0].page_content[:1000])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=[
        "\n\n",
        "\n",
        ". ",
        " ",
        ""
    ]
)
```

```python
chunks_1000 = text_splitter.split_documents(documents)

print("Documents before chunking:", len(documents))
print("Chunks after chunking:", len(chunks_1000))
```

```python
chunk = chunks_1000[0]

print("Metadata:")
print(chunk.metadata)

print("\nChunk length:")
print(len(chunk.page_content))

print("\nChunk content:")
print(chunk.page_content)
```

```python
for i in range(5):
    print("=" * 80)
    print(f"CHUNK {i}")
    
    print("\nSource:")
    print(chunks_1000[i].metadata.get("source_file"))
    
    print("Page:")
    print(chunks_1000[i].metadata.get("page_number"))
    
    print("Length:")
    print(len(chunks_1000[i].page_content))
    
    print("\nContent:")
    print(chunks_1000[i].page_content)
    
    print()
```
